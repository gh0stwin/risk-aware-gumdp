# Merged plot

````python
def add_box_plot(fig, data_to_plot, row, col, x_axis_lbl, y_axis_lbl, y_range=None):

    for erm_val, data in data_to_plot.items():
        xs = [str(erm_val)] * len(data["f_vals"])

        fig.add_trace(
            go.Box(
                y=np.array(data["f_vals"]),
                x=xs,
                showlegend=False,
                line=dict(color="#636efa"),
                offsetgroup='A'
            ),
            row=row,
            col=col
        )

    fig.update_xaxes(
        title_text=x_axis_lbl,
        type="category",
        categoryorder="array",
        categoryarray=[str(k) for k in data_to_plot.keys()],
        row=row,
        col=col,
        showline=True,
        linecolor="#aeb6c2",
        mirror=True,
        gridcolor="#e6ebeb",
    )

    fig.update_yaxes(showline=True, linecolor="#aeb6c2", mirror=True, gridcolor="#e6ebeb",)
    
    if col == 1:
        fig.update_yaxes(
            title_text=y_axis_lbl,
            row=row,
            col=col
        )

    if y_range:
        fig.update_yaxes(range=y_range, row=row, col=col)
````

````python
def add_linear_mdp_box_plot(fig, data_to_plot, row, col, x_axis_lbl, y_axis_lbl, y_range=None):
    
    for i, (erm_val, data) in enumerate(data_to_plot["BI"].items()):
        xs = [str(erm_val)] * len(data["f_vals"])
    
        fig.add_trace(go.Box(
            y=np.array(data["f_vals"]),
            x=xs,
            name="ERM-BI",                    # legend label
            legendgroup="ERM-BI",             # group legend entries
            showlegend=(i == 0),          # only show once
            line=dict(color="#ef553b"),
            offsetgroup="A"
        ))
    
    for i, (erm_val, data) in enumerate(data_to_plot["MCTS"].items()):
        xs = [str(erm_val)] * len(data["f_vals"])
    
        fig.add_trace(go.Box(
            y=np.array(data["f_vals"]),
            x=xs,
            name="ERM-MCTS",                  # legend label
            legendgroup="ERM-MCTS",
            showlegend=(i == 0),
            line=dict(color="#636efa"),
            offsetgroup="B"
        ))

    if y_range:
        fig.update_yaxes(range=y_range, row=row, col=col)
    
    #fig.update_layout(legend_title_text="Algorithm", boxmode="group", **default_layout)
    fig.update_xaxes(
        title_text=x_axis_lbl,
        showline=True,
        linecolor="#aeb6c2",
        mirror=True,
        gridcolor="#e6ebeb",
    )
    
    fig.update_yaxes(showline=True, linecolor="#aeb6c2", mirror=True, gridcolor="#e6ebeb",)

    fig.update_yaxes(
        title_text=y_axis_lbl,
        row=row,
        col=col
    )
    
````

````python
print(MERGED_PLOT_DATA.keys())

fig = make_subplots(
    rows=2,
    cols=3,
    subplot_titles=["Standard MDP", "Maximum state entropy expl.", "Imitation learning", "Multi-objective (weighted)", "Multi-objective (max)", "Multi-objective (min)"],
    horizontal_spacing=0.06,   # default ≈ 0.02
)

# Plot linear MDP data in position (1,1).
add_linear_mdp_box_plot(fig, MERGED_PLOT_DATA["linear_mdp"], row=1, col=1, x_axis_lbl="Beta", y_axis_lbl="Cost")

positions = [(1,1), (1,2), (1,3), (2,1), (2,2), (2,3)]
for (row, col), (setting, data) in zip(positions, MERGED_PLOT_DATA.items()):
    
    if setting == "linear_mdp":
        continue
    add_box_plot(fig, data, row=row, col=col, x_axis_lbl="Beta", y_axis_lbl="Cost")

default_layout['margin'] = dict(l=10, t=60, b=80, r=10)
fig.update_layout(
    boxmode="group",
    width=1100,
    height=620,
    annotations=[
        dict(
            font=dict(size=20)
        )
        for ann in fig.layout.annotations
    ],
    **default_layout
)

fig.update_layout(
    legend=dict(
        x=0.5,
        y=-0.18,
        xanchor="center",
        yanchor="top",
        font=dict(size=18),
        orientation="h"
    )
)

fig.write_image("imgs/merged_plot.pdf")
fig.show()
````
